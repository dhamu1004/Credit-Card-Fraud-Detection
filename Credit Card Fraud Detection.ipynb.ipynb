```python
import pandas as pd

data = {
    "Transaction Amount": [100, 25000, 500, 45000, 700],
}

df = pd.DataFrame(data)

def detect_fraud(amount):
    if amount > 10000:
        return "Fraud"
    return "Normal"

df["Status"] = df["Transaction Amount"].apply(detect_fraud)

print(df)

test_amount = 50000

print("\nTest Transaction:", test_amount)
print("Prediction:", detect_fraud(test_amount))
```
